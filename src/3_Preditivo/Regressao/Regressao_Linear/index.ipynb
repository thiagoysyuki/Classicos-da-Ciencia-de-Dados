# Regressão Linear 

O primeiro algoritmo de regressão que vamos explorar é a **Regressão Linear**. Acredito que seja o algoritmo de regressão mais conhecido e utilizado, pois exisitem inúmeros fenômenos que podem ser modelados por uma função linear.
Como químico de formação utilizei muito a regressão linear simples para criar curvas de calibração, onde a concentração de um analito é relacionada com a resposta do equipamento.
